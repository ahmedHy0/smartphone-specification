# Smartphone Data Analysis Project.

```python
import pandas as pd
import numpy as np

# Adding encoding='latin1' fixes the UnicodeDecodeError
df = pd.read_csv('Mobiles Dataset (2025).csv', encoding='latin1')

print(df.dtypes) # Displays data structure [cite: 38]
df.head(10)      # Displays first 10 rows [cite: 38]
```

# Section B: NumPy Array Programming - Data Normalization

```python
# Convert to numpy array
weight_array = df['Mobile Weight'].str.replace('g', '').astype(float).values

# Normalization formula: (x - min) / (max - min)
normalized_weight = (weight_array - np.min(weight_array)) / (np.max(weight_array) - np.min(weight_array))

# Add it back to the dataframe to show the result
df['Normalized_Weight'] = normalized_weight
df[['Mobile Weight', 'Normalized_Weight']].head()
```

# Section C: Sorting, Ranking, and Statistical Summary

```python
# Sorting by RAM and Ranking
df_sorted = df.sort_values(by='RAM', ascending=False)
df['RAM_Rank'] = df['RAM'].rank(ascending=False)

# Comprehensive statistical summary
df.describe()
```

Data Analysis:
* The statistical summary shows the distribution of smartphone specs across the dataset.
* For instance, the mean value of the weight column indicates the average weight of a modern smartphone in this dataset.
* The gap between the minimum and maximum values highlights the diversity between budget and flagship models

# Section D: Data Cleaning (Handling Missing Values and Duplicates)

```python
# 1. Handling Duplicates
df.drop_duplicates(inplace=True)

# 2. Safety Check & Column Extraction
# Using r'...' creates a raw string to fix the \d SyntaxWarning
if 'RAM_Val' not in df.columns:
    df['RAM_Val'] = df['RAM'].str.extract(r'(\d+)').astype(float)

# 3. Handling Missing Data (Imputation)
# Filling missing RAM values with the mode
df['RAM'] = df['RAM'].fillna(df['RAM'].mode()[0])

# 4. Data Transformation (Mapping)
# Cleaning brand names to be consistent
df['Company Name'] = df['Company Name'].map(lambda x: str(x).strip().title())

# 5. Binning (Categorizing)
# Creating performance tiers based on RAM capacity
df['Performance_Tier'] = pd.cut(df['RAM_Val'], bins=[0, 4, 8, 24], labels=['Budget', 'Mid', 'Flagship'])

# 6. Outliers
# Removing rows with invalid or 0 RAM values
df = df[df['RAM_Val'] > 0]

# Display results
df[['Company Name', 'RAM', 'Performance_Tier']].head()
```

Data Cleaning Logic:
* Missing Data: I used mode imputation for RAM because it is a categorical specification where the most frequent value is a safe statistical substitute.
* Outliers: Phones weighing less than 50g were removed as they likely represent data entry errors rather than actual smartphones.
* Duplicates: Removing duplicates ensures that specific models don't skew the statistical averages.
